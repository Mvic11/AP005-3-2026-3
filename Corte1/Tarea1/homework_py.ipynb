```python
a = input("Enter a number: ")  # Pide un texto al usuario para 'a'
a = int(a)  # Convierte 'a' a número entero
b = input("Enter b number: ")  # Pide un texto al usuario para 'b'
b = float(b)  # Convierte 'b' a número decimal (flotante)
c = a + b  # Suma 'a' y 'b' guardando el resultado en 'c'

if a == b:  # Comprueba si los valores numéricos son iguales
    print("equal")  # Imprime "equal" si valen lo mismo
else:  # Si los valores numéricos son distintos
    print("Different")  # Imprime "Different"

print("Type of a is: ", type(a))  # Muestra el tipo de dato de 'a' (int)
print("Type of b is: ", type(b))  # Muestra el tipo de dato de 'b' (float)
print("c = ", c)  # Muestra el valor final de la suma 'c'

if type(a) == type(b):  # Comprueba si 'a' y 'b' son del mismo tipo de dato
    print("a and b are of the same type")  # Imprime si coinciden en tipo
else:  # Si son de tipos de datos diferentes
    print("a and b are of different type")  # Imprime que sus tipos difieren
```
