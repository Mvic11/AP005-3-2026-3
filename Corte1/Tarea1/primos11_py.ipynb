```python
import time  # Importa el módulo 'time' para medir el tiempo de ejecución

inicio = time.time()  # Guarda el tiempo exacto de inicio en segundos

for i in range(1, 31):  # Recorre números del 1 al 30
    conta = 0  # Inicializa el contador de divisores para el número 'i'
    for n in range(1, i + 1):  # Recorre posibles divisores desde 1 hasta 'i'
        residue = i % n  # Obtiene el residuo de dividir 'i' entre 'n'
        if residue == 0:  # Comprueba si 'n' divide a 'i' exactamente
            conta = conta + 1  # Suma 1 al contador de divisores

    if conta == 2:  # Verifica si tiene solo 2 divisores
        print(f"{i} es un primo")  # Imprime el número primo
        print("\n")  # Imprime un salto de línea adicional

fin = time.time()  # Guarda el tiempo final en segundos
print("t = ", (fin - inicio) * 1000)  # Muestra el tiempo transcurrido en milisegundos
```
