```python
a = 1  # Inicializa la variable de control del bucle 'while' en 1
value = input("Ingrese un valor")  # Pide al usuario un número límite
value = int(value)  # Convierte la entrada ingresada a un número entero

while a == 1:  # Bucle principal que se ejecuta mientras 'a' sea igual a 1
    for i in range(1, value + 1):  # Recorre números desde el 1 hasta 'value'
        conta = 0  # Reinicia el contador de divisores para el número 'i'
        for n in range(1, i + 1):  # Prueba todos los posibles divisores 'n' hasta 'i'
            residue = i % n  # Calcula el residuo de la división entre 'i' y 'n'
            if residue == 0:  # Comprueba si 'n' divide a 'i' de forma exacta
                conta = conta + 1  # Suma 1 al contador si se encuentra un divisor

            # print("i = ", i)  # Comentario: Imprimiría el valor actual de 'i'
            # print("n = ", n)  # Comentario: Imprimiría el divisor actual 'n'
            # print("residue = ", residue)  # Comentario: Imprimiría el residuo
            # print("conta = ", conta)  # Comentario: Imprimiría los divisores acumulados
    if conta == 2:  # Evalúa la cantidad de divisores del ÚLTIMO número 'i'
        print(f"{i} es un primo")  # Imprime si el último número evaluado es primo
        print("\n")  # Imprime un salto de línea
    else:  # Si el contador final es distinto de 2
        print(f"{i} NOOO es un primo")  # Imprime que el último número NO es primo
        print("\n")  # Imprime un salto de línea

    print("Do you want to continue?. Press 1 to do that")  # Pregunta si desea repetir
    a = input()  # Recibe la opción ingresada por el usuario
    a = int(a)  # Convierte la opción elegida a entero

    if a != 1:  # Comprueba si la opción digitada es diferente de 1
        break  # Interrumpe y sale de inmediato del bucle 'while'

    value = input("Ingrese un valor")  # Pide un nuevo número si decidió continuar
    value = int(value)  # Convierte el nuevo número ingresado a entero
```
