```python
import time  # Importa el módulo 'time' para pausar la ejecución del programa

cadena = "Python"  # Guarda la palabra 'Python' en la variable 'cadena'

for letra in cadena:  # Recorre cada carácter de la palabra 'Python'
    if letra == "t":  # Comprueba si la letra actual es la 't'
        continue  # Salta la 't' y pasa de inmediato a la siguiente letra
    print(letra)  # Muestra en pantalla la letra actual
    time.sleep(1)  # Pausa la ejecución durante 1 segundo
```
