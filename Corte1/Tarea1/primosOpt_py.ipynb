```python
# 9) Imprimir los números primos existentes entre 0 y 30
tope_rango = 30  # Límite superior para buscar primos (hasta 29)
n = 0  # Inicializa el número a evaluar
primo = True  # Bandera que asume que el número es primo por defecto

while n < tope_rango:  # Repite la búsqueda mientras 'n' sea menor al tope
    for div in range(2, n):  # Prueba divisores desde 2 hasta 'n - 1'
        if n % div == 0:  # Si la división es exacta
            primo = False  # Cambia la bandera indicando que NO es primo
    if primo:  # Si la bandera se mantiene verdadera
        print(n)  # Imprime el número primo (nota: imprime 0 y 1 por falla del algoritmo)
    else:  # Si no fue primo
        primo = True  # Reinicia la bandera a True para el siguiente número
    n += 1  # Incrementa 'n' para evaluar el siguiente número


# 10) ¿Se puede mejorar el proceso del punto 9? Utilizar las sentencias break y/ó continue para tal fin
n = 0  # Reinicia el número a evaluar
primo = True  # Reinicia la bandera

while n < tope_rango:  # Bucle principal para evaluar números hasta el tope
    for div in range(2, n):  # Prueba divisores posibles
        if n % div == 0:  # Si encuentra un divisor exacto
            primo = False  # Marca el número como no primo
            break  # Interrumpe el bucle 'for' de inmediato al confirmar que no es primo
    if primo:  # Si conserva el estado de primo
        print(n)  # Imprime el número primo
    else:  # Si fue marcado como no primo
        primo = True  # Restablece la bandera a True para la siguiente iteración
    n += 1  # Incrementa 'n' en 1


# 11) En los puntos 9 y 10, se diseñó un código que encuentra números primos y además se lo optimizó. ¿Es posible saber en qué medida se optimizó?
ciclos_sin_break = 0  # Contador de iteraciones sin usar 'break'
n = 0  # Reinicia 'n'
primo = True  # Reinicia la bandera

while n < tope_rango:  # Bucle de prueba sin optimización
    for div in range(2, n):  # Evalúa todos los divisores sin detenerse
        ciclos_sin_break += 1  # Suma 1 al contador por cada iteración ejecutada
        if n % div == 0:  # Comprueba si es divisible
            primo = False  # Marca como no primo
    if primo:  # Si es primo
        print(n)  # Imprime el número
    else:  # Si no es primo
        primo = True  # Restablece la bandera
    n += 1  # Pasa al siguiente número

print("Cantidad de ciclos: " + str(ciclos_sin_break))  # Muestra el total de iteraciones sin 'break'


ciclos_con_break = 0  # Contador de iteraciones usando 'break'
n = 0  # Reinicia 'n'
primo = True  # Reinicia la bandera

while n < tope_rango:  # Bucle de prueba con optimización
    for div in range(2, n):  # Evalúa divisores
        ciclos_con_break += 1  # Suma 1 al contador por cada iteración
        if n % div == 0:  # Comprueba si es divisible
            primo = False  # Marca como no primo
            break  # Detiene la prueba de divisores inmediatamente
    if primo:  # Si es primo
        print(n)  # Imprime el número
    else:  # Si no es primo
        primo = True  # Restablece la bandera
    n += 1  # Pasa al siguiente número

print("Cantidad de ciclos: " + str(ciclos_con_break))  # Muestra el total de iteraciones con 'break'
print(
    "Se optimizó a un "
    + str(ciclos_con_break / ciclos_sin_break)
    + "% de ciclos aplicando break"
)  # Muestra la razón de rendimiento entre ambos métodos


# 12) Si la cantidad de números que se evalúa es mayor a treinta, esa optimización crece?
tope_rango = 100  # Aumenta el límite superior a 100
ciclos_sin_break = 0  # Reinicia contador sin 'break'
n = 0  # Reinicia 'n'
primo = True  # Reinicia la bandera

while n < tope_rango:  # Bucle sin 'break' hasta 100
    for div in range(2, n):  # Revisa todos los divisores posibles
        ciclos_sin_break += 1  # Incrementa el contador de iteraciones
        if n % div == 0:  # Si es divisible
            primo = False  # Marca como no primo
    if primo:  # Si es primo
        print(n)  # Imprime el número
    else:  # Si no es primo
        primo = True  # Restablece la bandera
    n += 1  # Pasa al siguiente número

print("Cantidad de ciclos: " + str(ciclos_sin_break))  # Imprime iteraciones totales sin 'break'

ciclos_con_break = 0  # Reinicia contador con 'break'
n = 0  # Reinicia 'n'
primo = True  # Reinicia la bandera

while n < tope_rango:  # Bucle con 'break' hasta 100
    for div in range(2, n):  # Revisa divisores
        ciclos_con_break += 1  # Incrementa el contador de iteraciones
        if n % div == 0:  # Si es divisible
            primo = False  # Marca como no primo
            break  # Sale del bucle 'for' al hallar el primer divisor
    if primo:  # Si es primo
        print(n)  # Imprime el número
    else:  # Si no es primo
        primo = True  # Restablece la bandera
    n += 1  # Pasa al siguiente número

print("Cantidad de ciclos: " + str(ciclos_con_break))  # Imprime iteraciones totales con 'break'
print(
    "Se optimizó a un "
    + str(ciclos_con_break / ciclos_sin_break)
    + "% de ciclos aplicando break"
)  # Muestra el porcentaje de ciclos resultantes comparado con el total
```
