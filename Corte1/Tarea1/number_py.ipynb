```python
import random  # Importa el módulo para generar números aleatorios
from matplotlib (
    import pyplot as plt,  # Importa 'pyplot' para la creación de gráficas
)

# Add your code below:  # Comentario del código original
numbers_a = range(1, 13)  # Genera una secuencia de números del 1 al 12
numbers_b = [
    random.randint(1, 1000) for i in range(12)
]  # Crea una lista de 12 enteros aleatorios entre 1 y 1000
plt.plot(
    numbers_a, numbers_b
)  # Grafica los datos de 'numbers_a' (eje X) contra 'numbers_b' (eje Y)
plt.show()  # Muestra en pantalla el gráfico generado
```
