```python
import time  # Importa el módulo 'time' para medir tiempos de ejecución

inicio = time.time()  # Registra el tiempo inicial en segundos

for i in range(0, 31):  # Recorre números desde el 0 hasta el 30
    conta = 0  # Inicializa el contador de divisores en 0 para cada número
    for n in range(1, i + 1):  # Prueba divisores desde 1 hasta 'i'
        residue = i % n  # Calcula el residuo de la división de 'i' entre 'n'
        if residue == 0:  # Comprueba si la división es exacta
            conta = conta + 1  # Incrementa el contador si 'n' es un divisor

    if conta == 2:  # Verifica si tiene exactamente 2 divisores (definición de número primo)
        print(f"{i} es un primo")  # Imprime que el número 'i' es primo

fin = time.time()  # Registra el tiempo final en segundos
print("t = ", (fin - inicio) * 1000)  # Muestra el tiempo total transcurrido en milisegundos
```
