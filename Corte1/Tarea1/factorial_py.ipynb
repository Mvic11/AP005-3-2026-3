```python
while True:  # Bucle infinito para pedir el valor continuamente
    value = int(input("Enter a positive integer value: "))  # Pide un texto y lo convierte a entero
    print("Value: ", value)  # Muestra el valor ingresado en pantalla
    a = isinstance(value, int)  # Verifica si 'value' es de tipo entero (devuelve True)
    if a == True and value > 0:  # Comprueba si es un entero positivo (mayor a 0)
        fact = 1  # Inicializa la variable del factorial en 1
        for i in range(1, value + 1):  # Recorre los números desde 1 hasta 'value'
            fact = fact * i  # Multiplica acumulativamente para calcular el factorial
        print(f"The factorial of {value} is: ", fact)  # Muestra el resultado del factorial
    else:  # Si el número no es mayor a 0
        print("Please, enter a positive integer number")  # Muestra mensaje de advertencia
```
