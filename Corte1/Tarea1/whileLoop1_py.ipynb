```python
for i in range(1, 6):  # Inicia el bucle exterior con 'i' tomando valores del 1 al 5
    while i <= 4:  # Repite el bucle mientras 'i' sea menor o igual a 4
        i += 1  # Incrementa el valor de 'i' en 1
        print(i)  # Imprime el nuevo valor de 'i'
    break  # Interrumpe y finaliza el bucle 'for' en su primera iteración
```
