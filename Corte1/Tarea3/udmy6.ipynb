Selección de Múltiples Columnas en Pandas

En este cuaderno se explora cómo seleccionar y extraer varias columnas simultáneamente de un `DataFrame` utilizando una lista de nombres de columnas dentro de los corchetes de selección (`df[["col1", "col2"]]`).

1. Carga de Datos y Muestra Inicial
Se realiza la importación de `pandas`, se carga el archivo de Excel `Nba Teams.xlsx` y se exploran sus primeros registros.

```python
import pandas as pd

# Carga del archivo de Excel a un DataFrame
nba = pd.read_excel("Nba Teams.xlsx")

# Muestra de las primeras 5 filas
nba.head()
```

2. Selección de un Subconjunto de Múltiples Columnas

Para seleccionar dos o más columnas a la vez, debemos pasar una **lista con los nombres de las columnas** dentro de los corchetes del DataFrame `df[["col1", "col2"]]` (notar los doble corchetes `[[ ]]`). El resultado obtenido es un nuevo DataFrame que contiene únicamente las columnas especificadas.

* **`nba[["Name", "Team"]]`**: Filtra el DataFrame devolviendo solo las columnas de **Nombre** y **Equipo**.
* **`nba[["Name", "Team", "Salary"]]`**: Filtra el DataFrame devolviendo las columnas de **Nombre**, **Equipo** y **Salario**.

```python
# Selección de 2 columnas: Name y Team
nba[["Name", "Team"]]

# Selección de 3 columnas: Name, Team y Salary
nba[["Name", "Team", "Salary"]]
```
