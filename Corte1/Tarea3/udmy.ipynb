Descarga y Exportación de Datos de COVID-19 en Nueva York

En este cuaderno se descargan datos sobre los casos de COVID-19 en la ciudad de Nueva York directamente desde la API en formato CSV del portal de datos abiertos de la ciudad, para posteriormente inspeccionarlos y guardarlos localmente en formato CSV.

1. Lectura e Inspección Inicial de Datos
Se utiliza `pandas.read_csv()` pasando una URL directa que contiene la fuente de datos en formato CSV.

```python
import pandas as pd

# URL del conjunto de datos en el portal de Datos Abiertos de NYC
url = "https://data.cityofnewyork.us/resource/rc75-m7u3.csv"

# Descarga y lectura del archivo CSV en un DataFrame
covid = pd.read_csv(url)

# Muestra de las primeras 5 filas del DataFrame
covid.head()
```

 2. Estructura y Diagnóstico del DataFrame (`covid.info()`)

Utilizamos el método `.info()` para consultar el resumen técnico del conjunto de datos, lo que nos permite verificar:
* El total de filas y columnas cargadas.
* Los nombres de los campos/columnas disponibles.
* La presencia de valores nulos o incompletos.
* Los tipos de datos asignados a cada variable (enteros, flotantes o texto).

```python
# Consulta técnica del conjunto de datos
covid.info()
```

3. Exportación de Datos a Archivos CSV (`to_csv`)

Guardamos la información del DataFrame en el almacenamiento local de la sesión de Colab mediante el método `.to_csv()`:

1. **`covid.to_csv('NYC_covid_cases.csv')`**: Exporta el conjunto de datos completo con todas sus columnas originales.
2. **`columns = [...]`**: Permite exportar únicamente un subconjunto de columnas específicas (`case_count`, `hospitalized_count`, `death_count`) para generar un archivo más liviano enfocado solo en el recuento de casos, hospitalizaciones y fallecimientos.

```python
# Guardar el DataFrame completo en un archivo CSV
covid.to_csv('NYC_covid_cases.csv')

# Guardar solo columnas seleccionadas en un nuevo archivo CSV
covid.to_csv('NYC1_covid_cases.csv', columns=['case_count', 'hospitalized_count', 'death_count'])
```
