 Unión y Combinación de Datos en Pandas (`pd.merge`)

En este cuaderno se explora el uso de combinaciones tipo SQL (*joins*: Inner, Left, Right y Outer) utilizando la función `pd.merge()` y el método `.merge()` de Pandas para integrar datos de clientes, ventas y productos.

 1. Carga Inicial de Datos
Se importan los archivos de clientes (`ABC - Customers.csv`) y ventas (`ABC - Sales.csv`) almacenados en Google Drive.

```python
import pandas as pd

# Carga de los conjuntos de datos principales
customers = pd.read_csv('/content/drive/MyDrive/udmy/dataAnalisys/ABC - Customers.csv')
sales = pd.read_csv('/content/drive/MyDrive/udmy/dataAnalisys/ABC - Sales.csv')

# Visualización del DataFrame de ventas
sales
```

2. Combinación Interna (*Inner Join*)

La combinación `how='inner'` conserva únicamente los registros cuyas claves (`cust_id`) coinciden en ambos DataFrames:
1. **Unión completa:** Une todas las columnas de `customers` y `sales`.
2. **Unión con selección de columnas:** Filtra columnas específicas del DataFrame de ventas antes de realizar la combinación para optimizar memoria.

```python
# Inner join completo
cust_sales = pd.merge(customers, sales, how='inner', on='cust_id')
cust_sales

# Inner join seleccionando columnas específicas de ventas
cust_sales = pd.merge(customers, sales[['cust_id', 'product', 'qty', 'store_location']], how='inner', on='cust_id')
cust_sales
```

3. Exploración de Identificadores Únicos e Izquierda (*Left Join*)

* **`nunique()`**: Cuenta el número de clientes únicos en cada tabla para identificar diferencias.
* **`how='left'` (*Left Join*):** Mantiene todos los registros de la tabla de la izquierda (`customers`), asignando valores nulos (`NaN`) a los clientes que no registraron compras en `sales`.
* **Identificación de clientes sin ventas:** Se filtran los clientes cuya columna `sale_id` resulta nula tras la combinación.

```python
# Conteo de clientes únicos en ambas tablas
print("Clientes únicos en customers:", customers['cust_id'].nunique())
print("Clientes únicos en sales:", sales['cust_id'].nunique())

# Left join
df1 = pd.merge(customers, sales, how='left', on='cust_id')
df1

# Verificación de valores nulos en ventas
df1['sale_id'].isnull()

# Filtrado de clientes que no han realizado ninguna compra
df1[df1['sale_id'].isnull()]['customer']
```

4. Combinaciones Derecha (*Right Join*) y Exterior (*Outer Join*)

* **`how='right'` (*Right Join*):** Conserva todos los registros del DataFrame derecho (`sales`), completando con nulos si faltan datos en `customers`.
* **`how='outer'` (*Outer Join*):** Incluye la unión completa de ambos DataFrames, conservando todas las filas de ambas tablas sin descartar información.

```python
# Right join
pd.merge(customers, sales, how='right', on='cust_id')

# Outer join
pd.merge(customers, sales, how='outer', on='cust_id')
```

5. Carga de Productos y Fusiones Encadenadas (*Multiple Merges*)

Se carga el catálogo de productos (`ABC - Products.csv`) y se realizan combinaciones encadenadas usando la sintaxis de método `.merge()`:
* **Fusión doble:** Une `customers` con `sales` por `cust_id`, y el resultado se une con `products` a través de `prod_id`.
* **Fusión encadenada con selección de columnas:** Permite incorporar únicamente el precio del producto sin arrastrar columnas innecesarias.

```python
# Carga del catálogo de productos
products = pd.read_csv('/content/drive/MyDrive/udmy/dataAnalisys/ABC - Products.csv')
products

# Fusión con sintaxis de función
pd.merge(customers, sales, how='inner', on='cust_id')

# Fusión encadenada con sintaxis de método (Clientes + Ventas)
customers.merge(sales, on='cust_id').head(2)

# Fusión triple encadenada (Clientes + Ventas + Productos completos)
customers.merge(sales, on='cust_id').merge(products, on='prod_id').head(2)

# Fusión triple encadenada (Clientes + Ventas + Precio de Producto)
customers.merge(sales, on='cust_id').merge(products[['prod_id', 'price']], on='prod_id').head(2)
```
