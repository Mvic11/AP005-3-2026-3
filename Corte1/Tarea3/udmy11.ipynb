Reemplazo e Imputación de Valores Nulos (`fillna`)

En este cuaderno se aborda cómo tratar los valores nulos o faltantes (`NaN`) reemplazándolos con valores específicos mediante el método `.fillna()`, explorando la aplicación global, por columnas individuales y el uso de `inplace = True`.

1. Conexión a Drive, Carga e Inspección Inicial
Montamos la unidad de Google Drive, importamos `pandas` y leemos el archivo de Excel `Nba Teams.xlsx` para observar sus primeros 10 registros.

```python
import pandas as pd
from google.colab import drive

# Montaje de Google Drive
drive.mount('/content/drive')

# Carga del conjunto de datos desde Excel
nba = pd.read_excel('/content/drive/MyDrive/udmy/dataAnalisys/Nba Teams.xlsx')

# Visualización de las primeras 10 filas
nba.head(10)
```

2. Reemplazo Global de Nulos (`nba.fillna(0)`)

El método `nba.fillna(0)` busca **todos los valores nulos** en cualquier columna del DataFrame y los reemplaza por `0`.

> **Nota:** Al ejecutarse sin `inplace = True`, este cambio se muestra temporalmente en pantalla y no modifica la variable original `nba`.

```python
# Reemplaza temporalmente todos los valores NaN por 0 en todo el DataFrame
nba.fillna(0)
```

3. Reemplazo de Nulos en Columnas Específicas

Para no alterar todo el conjunto de datos, podemos imputar valores faltantes únicamente en columnas seleccionadas:

* **`nba[["College"]].fillna(0, inplace = True)`**: Intenta reemplazar los nulos de la columna `College` por `0`.
* **`nba["Name"].fillna("Missing name", inplace = True)`**: Reemplaza permanentemente cualquier valor nulo de la columna `Name` por la etiqueta en texto `"Missing name"`.
