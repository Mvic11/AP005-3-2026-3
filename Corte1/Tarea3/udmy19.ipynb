Filtrado Multicondicional con el Operador Lógico AND (`&`)

En este cuaderno se analiza cómo filtrar un `DataFrame` aplicando múltiples condiciones simultáneas utilizando el operador bit a bit `&` (AND lógico). Bajo este operador, una fila solo se incluye en el resultado si **todas** las condiciones se cumplen al mismo tiempo (`True`).

1. Conexión a Drive e Importación Optimizada
Se realiza el montaje de Google Drive, la carga del conjunto de datos parseando las columnas de fecha (`parse_dates`) y la asignación de los tipos categórico y booleano.

```python
import pandas as pd
from google.colab import drive

# Montaje de Google Drive
drive.mount('/content/drive')

# Carga e inicialización optimizada de datos
insta = pd.read_csv(
    '/content/drive/MyDrive/udmy/dataAnalisys/Instagram.csv',
    parse_dates=["reg date", "last login"]
)

insta["gender"] = insta["gender"].astype("category")
insta["Country"] = insta["Country"].astype("category")
insta["Influencer"] = insta["Influencer"].astype("bool")

# Visualización de las primeras filas
insta.head()
```

 2. Filtrado Múltiples Condiciones en una Sola Máscara

Para combinar varias expresiones lógicas en una sola línea de código, cada condición individual **debe ir encerrada entre paréntesis `()`**:

* **Condición 1**: `insta["gender"] == "Male"` (Género masculino).
* **Condición 2**: `insta["reg date"] > "2015-12-31"` (Registrados a partir del año 2016).
* **Condición 3**: `insta["Country"] == "United States"` (País de origen Estados Unidos).

```python
# Creación de una máscara compuesta con operador &
mask = (insta["gender"] == "Male") & (insta["reg date"] > "2015-12-31") & (insta["Country"] == "United States")

# Aplicación del filtro
insta[mask]
```

 3. Modularización de Máscaras e Identificación de Sensibilidad a Mayúsculas

Dividir las condiciones en variables independientes (`mask1`, `mask2`, `mask3`) facilita la lectura y el mantenimiento del código al momento de realizar la combinación lógica `insta[mask1 & mask2 & mask3]`.

> **Atención con los valores de texto**: Las comparaciones de cadenas en Pandas son estrictas con las mayúsculas/minúsculas. Evaluar `insta["gender"] == "male"` (en minúscula) no coincidirá con `"Male"` y devolverá un `DataFrame` vacío.

```python
# Construcción modular de condiciones por separado
mask1 = insta["gender"] == "Male"
mask2 = insta["reg date"] > "2015-12-31"
mask3 = insta["Country"] == "United States"

# Unión de máscaras modulares
insta[mask1 & mask2 & mask3]

# Ejemplo de inconsistencia por minúsculas (devolverá 0 filas si el dato original es "Male")
mask_error = (insta["gender"] == "male") & (insta["reg date"] > "2015-12-31")
insta[mask_error]
```
