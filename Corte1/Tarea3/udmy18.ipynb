 Filtrado Condicional de Datos (MÁSCARAS BOOLEANAS)

En este cuaderno se analiza cómo realizar filtrados de datos basados en condiciones lógicas sobre las columnas de un `DataFrame`. Además, se optimiza la carga inicial del archivo parseando fechas automáticamente desde `pd.read_csv()`.

 1. Conexión a Drive e Importación Optimizada (`parse_dates`)
En lugar de convertir las fechas manualmente después de cargar el archivo, utilizamos el argumento `parse_dates` en `pd.read_csv()` para transformar las columnas `reg date` y `last login` a tipo `datetime64` desde el momento de la lectura.

```python
import pandas as pd
from google.colab import drive

# Montaje de Google Drive
drive.mount('/content/drive')

# Carga optimizada del archivo definiendo el parseo de fechas al vuelo
insta = pd.read_csv(
    '/content/drive/MyDrive/udmy/dataAnalisys/Instagram.csv',
    parse_dates=["reg date", "last login"]
)

# Conversión a tipos de datos optimizados
insta["gender"] = insta["gender"].astype("category")
insta["Country"] = insta["Country"].astype("category")
insta["Influencer"] = insta["Influencer"].astype("bool")

# Verificación de la estructura y de las primeras 3 filas
insta.info()
insta.head(3)
```

2. Filtrado Directo por Igualdad (`==`)

Para filtrar filas en Pandas se pasa una condición lógica dentro de corchetes `df[condicion]`. Esto evalúa cada fila y devuelve únicamente aquellas que cumplen la condición (`True`):

* **`insta[insta["gender"] == "Male"]`**: Filtra y muestra solo los usuarios de género masculino.
* **`insta[insta["Country"] == "Australia"]`**: Filtra los registros pertenecientes a Australia.

```python
# Filtrado de usuarios masculinos
insta[insta["gender"] == "Male"]

# Filtrado de usuarios localizados en Australia
insta[insta["Country"] == "Australia"]
```

3. Filtrado con Máscaras Booleanas Guardadas y Evaluaciones de Desigualdad (`!=` / `bool`)

Es una buena práctica almacenar las condiciones complejas en variables separadas (llamadas habitualmente **máscaras** o `mask`) para mejorar la legibilidad del código:

* **`mask = insta["gender"] == "Female"`**: Almacena la serie booleana y la aplica posteriormente (`insta[mask]`).
* **Columnas Booleanas Directas**: Si una columna ya es de tipo `bool` (como `Influencer`), no es necesario usar `== True`. Pasar la columna directamente `insta[insta["Influencer"]]` extrae todos los registros verdaderos.
* **Operador de Desigualdad (`!=`)**: `insta["Country"] != "United States"` selecciona todos los registros excepto los provenientes de Estados Unidos.

```python
# Uso de una variable máscara para filtrar
mask = insta["gender"] == "Female"
insta[mask]

# Filtrado directo por columna booleana
insta[insta["Influencer"]]

# Filtrado por desigualdad (todos los países excepto Estados Unidos)
mask1 = insta["Country"] != "United States"
insta[mask1]
```
