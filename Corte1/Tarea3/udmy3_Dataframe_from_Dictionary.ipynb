Creación de DataFrames a partir de Diccionarios en Pandas

En este cuaderno se explora la construcción de un `DataFrame` de Pandas utilizando un diccionario de Python (`dict`), donde cada clave (*key*) representa el nombre de una columna y cada lista asociada representa los datos de esa columna.

1. Importación de Librerías y Definición del Diccionario
Comenzamos definiendo un diccionario `data` con la información sobre nombres, edades y géneros de tres personas.

```python
import pandas as pd

# Definición del diccionario de datos
data = {
    "name": ["Sally", "Jhon", "Kim"],
    "age": [4, 6, 7],
    "gender": ["female", "male", "male"]
}
```

2. Conversión a DataFrame y Personalización de Índices

* **`pd.DataFrame(data)`**: Convierte el diccionario directamente en un DataFrame. Las claves (`"name"`, `"age"`, `"gender"`) se convierten en las columnas, y Pandas asigna índices numéricos automáticos (0, 1, 2).
* **`index = ["Kid1", "Kid2", "Kid3"]`**: Reemplaza el índice numérico por defecto asignando identificadores personalizados a cada fila.

```python
# Creación básica del DataFrame desde el diccionario
pd.DataFrame(data)

# Creación del DataFrame asignando etiquetas personalizadas a los índices
pd.DataFrame(data, index=["Kid1", "Kid2", "Kid3"])
```
