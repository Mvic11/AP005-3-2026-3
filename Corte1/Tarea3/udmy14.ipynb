Ordenamiento de Datos en Pandas (`sort_values`)

En este cuaderno se explora el método `.sort_values()` para ordenar un `DataFrame` según los valores de una o varias columnas, modificando el criterio de orden (ascendente/descendente) y la posición de los datos nulos.

 1. Conexión a Drive e Importación de Datos
Se realiza el montaje de Google Drive, la importación de `pandas` y la lectura del archivo de Excel `Nba Teams.xlsx`.

```python
import pandas as pd
from google.colab import drive

# Montaje de Google Drive
drive.mount('/content/drive')

# Carga del conjunto de datos desde Excel
nba = pd.read_excel('/content/drive/MyDrive/udmy/dataAnalisys/Nba Teams.xlsx')

# Visualización de los primeros registros
nba.head()
```

 2. Ordenamiento por una Sola Columna

El método `.sort_values()` ordena las filas del DataFrame según la columna indicada:

* **`nba.sort_values("Name").head(20)`**: Ordena alfabéticamente por la columna `Name` de la A a la Z (ascendente por defecto) y muestra los primeros 20 registros.
* **`nba.sort_values("Salary")`**: Ordena los datos de menor a mayor salario.
* **`ascending = False`**: Invierte el orden para mostrar primero los salarios más altos (de mayor a menor).

```python
# Ordenamiento alfabético por Nombre (primeros 20)
nba.sort_values("Name").head(20)

# Ordenamiento por Salario (ascendente por defecto)
nba.sort_values("Salary")

# Ejemplo para ordenar de mayor a menor salario (guardando los cambios con inplace)
# nba.sort_values("Salary", ascending=False, inplace=True)
```

 3. Ordenamiento por Múltiples Columnas y Manejo de Nulos (`na_position`)

* **Múltiples columnas `["Team", "Name"]`**: Ordena los datos jerárquicamente. Primero agrupa los registros por el nombre del equipo (`Team`) en orden alfabético y, en caso de empate dentro del mismo equipo, ordena a los jugadores por su nombre (`Name`).
* **Parámetro `na_position`**: Por defecto los valores nulos (`NaN`) se ubican al final (`na_position = 'last'`). Al especificar `na_position = 'first'`, los registros con datos faltantes se ubican al inicio del DataFrame.

```python
# Ordenamiento jerárquico por Equipo y luego por Nombre
nba.sort_values(["Team", "Name"])

# Ordenamiento por múltiples columnas posicionando los valores nulos al principio
# nba.sort_values(["Team", "Name"], na_position='first')
```
