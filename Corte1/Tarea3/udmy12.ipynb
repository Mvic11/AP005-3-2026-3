Operaciones Aritméticas en Series de Pandas

En este cuaderno se explora cómo realizar operaciones matemáticas (suma, resta, multiplicación y división) sobre columnas de tipo `Series` en Pandas, utilizando tanto operadores aritméticos estándar (`+`, `-`, `*`, `/`) como métodos integrados (`.add()`, `.sub()`, `.mul()`, `.div()`).

1. Conexión a Drive e Importación de Datos
Se realiza el montaje de Google Drive, la importación de `pandas` y la lectura de la columna `Salary` del archivo `Nba Teams.xlsx`.

```python
import pandas as pd
from google.colab import drive

# Montaje de Google Drive
drive.mount('/content/drive')

# Carga del conjunto de datos desde Excel
nba = pd.read_excel('/content/drive/MyDrive/udmy/dataAnalisys/Nba Teams.xlsx')

# Visualización de la columna Salary
nba["Salary"]
```

2. Operaciones de Suma y Creación de Nuevas Columnas

Podemos sumar valores constantes a cada fila de una columna de dos formas:
* **Método `.add(valor)`**: `nba["Salary"].add(50000)` suma 50,000 a cada salario.
* **Operador `+`**: `nba["Salary"] + 2000000` suma 2,000,000 a cada salario.
* **Asignación de columna**: `nba["bonus_salary"] = nba["Salary"] + 30000` calcula el salario con bono de 30,000 y lo almacena en una nueva columna.

```python
# Suma mediante el método .add()
nba["Salary"].add(50000)

# Suma mediante el operador +
nba["Salary"] + 2000000

# Creación de una nueva columna con la suma resultante
nba["bonus_salary"] = nba["Salary"] + 30000
nba
```

3. Resta, División y Conversión de Unidades (Multiplicación)

* **Resta (`.sub()` o `-`)**: `nba["Salary"].sub(55000)` resta 55,000 a cada salario.
* **División (`.div()` o `/`)**: `nba["Salary"].div(1000000)` divide el salario entre 1,000,000 para expresarlo en millones de dólares.
* **Multiplicación (`.mul()` o `*`)**: `nba["Weight"].mul(0.453592)` multiplica el peso expresado en libras por `0.453592` para convertirlo a kilogramos.

```python
# Resta de 55,000 a la columna de salarios
nba["Salary"].sub(55000)

# División para convertir salarios a millones
nba["Salary"].div(1000000)

# Multiplicación para convertir el peso de libras a kilogramos
nba["Weight"].mul(0.453592)
```
