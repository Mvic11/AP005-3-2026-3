El método `.info()` devuelve un resumen general de la tabla, incluyendo:
* El número total de filas y columnas.
* Los nombres de las columnas y la cantidad de valores no nulos en cada una.
* El tipo de datos almacenado en cada columna (`int64`, `float64`, `object`, etc.).
* La cantidad de memoria RAM utilizada por el DataFrame.
