Carga con Encadenamiento de Métodos y Manejo de Nulos (`fillna`)

En este cuaderno se explora cómo aplicar limpieza directa al momento de importar un conjunto de datos encadenando el método `.dropna()`, además de realizar la imputación de valores faltantes en columnas específicas como la altura (`Height`).

1. Conexión a Drive e Importación con Filtrado Directo
Se conecta Google Drive y se realiza la lectura del archivo `Nba Teams.xlsx` aplicando de forma inmediata `.dropna(how = "all")` para descartar desde el inicio cualquier fila que esté completamente vacía.

```python
import pandas as pd
from google.colab import drive

# Montaje de Google Drive
drive.mount('/content/drive')

# Carga del conjunto de datos filtrando filas completamente vacías al vuelo
nba = pd.read_excel("/content/drive/MyDrive/udmy/dataAnalisys/Nba Teams.xlsx").dropna(how="all")

# Visualización del DataFrame cargado
nba
```

2. Inspección General e Imputación de Valores Faltantes en Cadenas

* **`nba.info()`**: Muestra la estructura de los datos cargados y el conteo de valores no nulos por columna.
* **`nba["Height"].fillna("Missing")`**: Selecciona la columna `Height` y reemplaza los datos faltantes (`NaN`) por la cadena de texto `"Missing"`.

> **Nota:** Dado que no se especifica `inplace=True` ni se asigna de nuevo a la columna (`nba["Height"] = ...`), este reemplazo se muestra como una vista temporal y no altera permanentemente el `DataFrame`.

```python
# Inspección técnica del DataFrame
nba.info()

# Reemplazo temporal de nulos en la columna de altura
nba["Height"].fillna("Missing")
```
