Eliminación de Columnas en Pandas (`drop` e `inplace`)

En este cuaderno se explora el uso del método `.drop()` para eliminar una o varias columnas de un `DataFrame`, comprendiendo la diferencia entre realizar una eliminación temporal y una permanente mediante el parámetro `inplace = True`.

1. Carga de Datos y Muestra Inicial
Importamos la librería `pandas`, cargamos el archivo de Excel `Nba Teams.xlsx` y visualizamos sus primeros registros.

```python
import pandas as pd

# Carga del conjunto de datos desde un archivo Excel
nba = pd.read_excel("Nba Teams.xlsx")

# Muestra de las primeras 5 filas
nba.head()
```

2. Eliminación Temporal de una Columna

Al utilizar `nba.drop("Salary", axis = 1)` sin el argumento `inplace = True`, Pandas retorna una copia del DataFrame sin la columna especificada, pero **no modifica la variable original** `nba`.

* **`"Salary"`**: Nombre de la columna a remover.
* **`axis = 1`**: Especifica que la eliminación se aplica sobre el eje de columnas.

```python
# Muestra una copia del DataFrame sin la columna 'Salary'
nba.drop("Salary", axis=1)

# Al imprimir 'nba', la columna 'Salary' sigue existiendo en el DataFrame original
nba
```

3. Eliminación Permanente usando `inplace = True`

Para hacer efectiva la eliminación sobre la variable `nba` sin necesidad de reasignarla (`nba = nba.drop(...)`), se utiliza el parámetro `inplace = True`.

```python
# Eliminación permanente de la columna 'Salary'
nba.drop("Salary", axis=1, inplace=True)

# Confirmación de que la columna 'Salary' ha sido removida definitivamente
nba
```

4. Eliminación de Múltiples Columnas a la Vez

El método `.drop()` también permite borrar varias columnas de forma simultánea pasando una **lista con sus nombres** `["Team", "Division"]`.

```python
# Eliminación permanente de múltiples columnas
nba.drop(["Team", "Division"], axis=1, inplace=True)

# Muestra del DataFrame final resultante
nba
```
