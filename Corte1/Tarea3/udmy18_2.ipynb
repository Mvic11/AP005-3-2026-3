 Combinación de DataFrames en Pandas (`pd.merge`)

En este cuaderno se explora el uso de `pd.merge()` y del método `.merge()` para unificar múltiples fuentes de datos (relaciones de tablas) sobre columnas clave en común (`student_id` y `course_id`), aplicando además agrupamientos y filtros sobre el resultado unificado.

1. Conexión a Drive, Carga de Archivos y Revisión Inicial
Se conecta Google Drive y se realiza la lectura de las tres tablas del sistema académico: **Cursos**, **Calificaciones** y **Estudiantes**.

```python
import pandas as pd
from google.colab import drive

# Montaje de Google Drive
drive.mount('/content/drive')

# Carga de las tres tablas relacionales desde CSV
courses = pd.read_csv('/content/drive/MyDrive/udmy/dataAnalisys/COLLEGE - Courses.csv')
grades = pd.read_csv('/content/drive/MyDrive/udmy/dataAnalisys/COLLEGE - Grades.csv')
students = pd.read_csv('/content/drive/MyDrive/udmy/dataAnalisys/COLLEGE - Students.csv')

# Inspección inicial de las tablas
courses.head()
grades.head()
students.head()
```

 2. Combinación de Tablas (`pd.merge`) y Filtrado

Para cruzar dos DataFrames basados en una columna en común (llave primaria / foránea), utilizamos `pd.merge()` especificando el parámetro `on`:

* **`pd.merge(students, grades, on = 'student_id')`**: Unifica la información personal de los estudiantes con sus calificaciones utilizando `student_id`.
* **Filtros Booleanos**: Una vez unificadas las tablas, se pueden aplicar filtros condicionales para buscar registros de estudiantes específicos por su ID (`df['student_id'] == 2332`).

```python
# Combinación básica y ordenamiento por calificación de mayor a menor
df = pd.merge(students, grades, on='student_id').sort_values('grade', ascending=False)
df.head(10)

# Consulta de registros para un estudiante específico
df[df['student_id'] == 2332]
df[df['student_id'] == 2368]
```

3. Encadenamiento de Uniones Relacionales (Múltiples Merges)

Podemos encadenar uniones para consolidar información de más de dos tablas en un solo DataFrame unificado:

1. Primero combinamos `students` con `grades` mediante `student_id`.
2. Luego unimos el resultado obtenido con `courses` mediante `course_id`.

```python
# Primera combinación: Estudiantes + Calificaciones
students_rec = students.merge(grades, on='student_id')

# Segunda combinación: Agregar la información de los Cursos
students_rec1 = students_rec.merge(courses, on='course_id')

# Inspección del DataFrame consolidado final
students_rec1.head()
```

4. Agrupamiento e Inspección por Curso e Historial de Calificaciones

Con la tabla totalmente consolidada, es posible realizar análisis de frecuencia y agrupamientos combinados mediante `.groupby()`:

* **`reg_course.size().sort_values()`**: Muestra la cantidad total de estudiantes inscritos por materia ordenados de menor a mayor.
* **`groupby(['course_name', 'last_name', 'first_name', 'grade'])`**: Agrupa el detalle del historial académico organizando materias, estudiantes y notas finales.

```python
# Conteo de inscripciones por materia
reg_course = students_rec1.groupby('course_name')
reg_course.size().sort_values()

# Detalle agrupado de estudiantes y notas por curso
students_records = students_rec1.groupby(['course_name', 'last_name', 'first_name', 'grade'])
students_records.size()
```
