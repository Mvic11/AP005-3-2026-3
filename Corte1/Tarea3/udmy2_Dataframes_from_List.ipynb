Creación de DataFrames a partir de Listas en Pandas

En este cuaderno se explora cómo construir un `DataFrame` de Pandas desde cero utilizando listas simples y listas anidadas (listas de listas), así como la personalización de los nombres de columnas y las etiquetas de los índices.

1. Importación de la Librería y Creación de un DataFrame Simple
Comenzamos importando la librería `pandas`. A partir de una lista simple de números (`[1, 2, 3, 4, 5, 6]`), Pandas genera un DataFrame de una sola columna asignando índices numéricos automáticos (0 a 5).

```python
import pandas as pd

# Lista unidimensional de números
numbers = [1, 2, 3, 4, 5, 6]

# Conversión a DataFrame
pd.DataFrame(numbers)
```

2. Creación de un DataFrame Multicolumna desde una Lista de Listas

Para representar datos multidimensionales (como filas y columnas), utilizamos una lista de listas (`data`), donde cada sublista representa un registro individual con nombre y edad.

Al no especificar encabezados, Pandas asigna nombres numéricos por defecto a las columnas (`0`, `1`).

```python
# Lista de listas con nombres y edades
data = [["Jhon", 23], ["Mary", 18], ["Jhon", 31]]

# Conversión básica a DataFrame
pd.DataFrame(data)
```

3. Personalización de Nombres de Columnas (`columns`) y Filtro de Índices (`index`)

Podemos asignar nombres descriptivos a las columnas y personalizar los identificadores de fila (índices):

* **`columns = ["name", "age"]`**: Asigna los nombres `"name"` y `"age"` a los encabezados de las columnas.
* **`index = ["id1", "id2", "id3"]`**: Reemplaza el índice numérico por defecto (0, 1, 2) por identificadores personalizados (`id1`, `id2`, `id3`).

```python
# Asignación de nombres a las columnas
pd.DataFrame(data, columns=["name", "age"])

# Asignación de nombres a las columnas y etiquetas personalizadas al índice
pd.DataFrame(data, columns=["name", "age"], index=["id1", "id2", "id3"])
```
