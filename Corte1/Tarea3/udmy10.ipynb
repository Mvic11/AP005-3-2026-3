Tratamiento de Valores Nulos en Pandas (`dropna`)

En este cuaderno se explora la limpieza de datos eliminando registros que contienen valores nulos o faltantes (`NaN`) mediante el método `.dropna()`, revisando sus distintas configuraciones (`how`, `axis` y `subset`).

1. Conexión a Drive, Importación e Inspección Inicial
Conectamos Google Drive, importamos `pandas` y revisamos los primeros y últimos registros del DataFrame para detectar la presencia de datos incompletos.

```python
import pandas as pd
from google.colab import drive

# Montaje de Google Drive
drive.mount('/content/drive')

# Carga del conjunto de datos desde Excel
nba = pd.read_excel("/content/drive/MyDrive/udmy/AP005320241/Nba Teams.xlsx")

# Inspección de los primeros 10 y últimos 5 registros
nba.head(10)
nba.tail()
```

2. Eliminación de Filas con Valores Nulos

El método `.dropna()` elimina por defecto cualquier fila que contenga al menos un valor nulo (`NaN`):
* Se puede guardar el resultado filtrado en una nueva variable (`nbaNoNull = nba.dropna()`).
* Si se ejecuta `nba.dropna()` sin el parámetro `inplace=True`, el DataFrame original `nba` permanece intacto.

```python
# Crear un nuevo DataFrame filtrado sin valores nulos
nbaNoNull = nba.dropna()
nbaNoNull

# Demostración: 'nba' conserva sus filas originales con nulos
nba.info()
```

3. Filtrado por Condición de Nulos (`how="all"`)

Por defecto, `.dropna()` elimina la fila si **cualquier** celda es nula (`how="any"`). Con el parámetro `how="all"`, solo se eliminarán las filas donde **absolutamente todos** los campos de esa fila sean nulos.

```python
# Elimina únicamente las filas que estén completamente vacías
nba.dropna(how="all")
```

4. Eliminación de Columnas con Nulos (`axis=1`)

Si cambiamos el eje a `axis=1` (o `axis="columns"`), `.dropna()` eliminará **columnas completas** si estas contienen al menos un valor nulo.

```python
# Elimina las columnas que tengan al menos un valor nulo
nba.dropna(axis=1)
```

5. Filtrado por Columnas Específicas (`subset`)

El parámetro `subset` permite restringir la evaluación de nulos a una lista concreta de columnas:
* **`subset = ["Name", "Salary", "College"]`**: Elimina la fila únicamente si falta el dato en alguno de estos tres campos, ignorando si hay nulos en otras columnas del DataFrame.

```python
# Elimina filas que tengan nulos en las columnas especificados
nba.dropna(subset=["Name", "Salary", "College"])
```
