Análisis Exploratorio de Datos: Frutas (`Fruits.csv`)

En este cuaderno se trabaja con un conjunto de datos referente a precios y tipos de frutas. Se abordan conceptos clave de Pandas como indices multinivel (MultiIndex), ordenamiento, operaciones entre columnas y manipulación de estructura (*stack/unstack*).

1. Carga e Inspección Inicial
Se realiza la importación de la librería `pandas` y la lectura inicial del archivo CSV almacenado en Google Drive.

```python
import pandas as pd

# Carga del conjunto de datos desde Google Drive
df = pd.read_csv("/content/drive/MyDrive/udmy/dataAnalisys/Fruits.csv")

# Muestra de las primeras 10 filas del DataFrame
df.head(10)
```

 2. Diagnóstico de la Estructura de Datos (`df.info()`)

El método `df.info()` permite verificar el número de registros, los tipos de datos en cada columna y si existen valores nulos:
* **Entradas totales:** 76 filas (del índice 0 al 75).
* **Columnas:** 3 (`type`, `color`, `price usd`).
* **Tipos de datos:** Texto (`object`) para tipo y color; decimal (`float64`) para el precio.

```python
df.info()
```

 3. Creación de un Índice Multinivel (MultiIndex)

Podemos transformar las columnas `type` y `color` en un índice compuesto de dos niveles.
* `df.set_index(["type", "color"])`: Asigna las columnas como índice.
* `inplace = True`: Aplica los cambios de forma permanente sobre el DataFrame `df`.

```python
# Aplicación del indice multinivel
df.set_index(["type", "color"], inplace=True)
df
```

4. Inspección de los Niveles del Índice

Con la propiedad `.index` y el método `get_level_values()` es posible explorar los nombres de las filas en cada nivel:
* **Nivel 0:** Corresponde a `type` (tipo de fruta).
* **Nivel 1:** Corresponde a `color` (color de la fruta).

```python
# Muestra la estructura general del MultiIndex
print(df.index)

# Extrae los valores del primer nivel (type)
print(df.index.get_level_values(0))

# Extrae los valores del segundo nivel (color)
print(df.index.get_level_values(1))
```

5. Definición Directa del Índice al Cargar el Archivo

En lugar de usar `set_index()`, podemos definir las columnas que actuarán como índice directamente en la función `pd.read_csv()` usando el argumento `index_col`.

```python
# Carga e indexación simultánea
df = pd.read_csv("/content/drive/MyDrive/udmy/dataAnalisys/Fruits.csv", index_col=["type", "color"])
df.head(10)
```

6. Ordenamiento por Índice (`sort_index`)

Al trabajar con índices complejos, ordenarlos mejora el rendimiento de las búsquedas y visualizaciones:
* `df.sort_index()`: Ordena alfabéticamente por todos los niveles.
* `ascending = False`: Ordena en sentido descendente.
* `level = N`: Permite especificar qué nivel concreto (0 o 1) se desea ordenar.

```python
# Orden alfabético predeterminado
df.sort_index()

# Orden descendente
df.sort_index(ascending=False)

# Ordenar únicamente por el Nivel 0 (type)
df.sort_index(level=0)

# Ordenar únicamente por el Nivel 1 (color)
df.sort_index(level=1)
```

7. Conversión y Creación de Nuevas Columnas

Calculamos una nueva columna `price in pounds` multiplicando la columna `price usd` por el factor de conversión `0.71`.

```python
# Selección de la columna de precio original
df["price usd"]

# Conversión de divisa y asignación a una nueva columna
df["price in pounds"] = df["price usd"] * 0.71
df
```

8. Selección de Datos con `.loc` en Índices Multinivel

El indexador `.loc` permite hacer filtrados avanzados utilizando la tupla del índice:
* `df.loc["Apples"]`: Filtra todas las filas del primer nivel donde el tipo sea "Apples".
* `df.loc["Apples", "Green"]`: Filtra especificando nivel 0 ("Apples") y nivel 1 ("Green").
* `df.loc[("Apples", "Green"), "price usd"]`: Selecciona un dato escalar especificando el índice completo y la columna deseada.

```python
# Filtrado por nivel 0
df.loc["Apples"]

# Filtrado por tupla (nivel 0, nivel 1)
df.loc["Apples", "Green"]

# Acceso a una celda específica indicando fila y columna
df.loc[("Apples", "Green"), "price usd"]
```

9. Reestructuración de Datos con `stack()` y `unstack()`

* **`stack()`**: Apila las columnas del DataFrame para convertirlas en un nivel adicional dentro de las filas (crea una Serie/DataFrame más "largo").
* **`to_frame()`**: Convierte una Serie resultante en un DataFrame de Pandas.
* **`unstack()`**: Proceso inverso; desplaza el último nivel del índice de filas hacia las columnas (vuelve a expandir el cuadro a una forma tabular horizontal).

```python
# Apilar columnas como parte del índice
df.stack()

# Convertir el resultado a un nuevo DataFrame df2
df2 = df.stack().to_frame()
df2

# Desapilar el nivel más profundo de filas devolviéndolo a columnas
df2.unstack()
```
