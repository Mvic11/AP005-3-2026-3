Selección y Acceso a Columnas en Pandas

En este cuaderno se explora cómo cargar un archivo de Excel y las distintas formas que ofrece Pandas para seleccionar y acceder a las columnas de un `DataFrame` (utilizando corchetes `[]` y sintaxis de atributos con punto `.`).

1. Carga de Datos y Muestra Inicial
Se realiza la importación de `pandas`, se carga el archivo `Nba Teams.xlsx` y se visualizan los primeros registros con el método `.head()`.

```python
import pandas as pd

# Carga del archivo Excel a un DataFrame
nba = pd.read_excel("Nba Teams.xlsx")

# Muestra de las primeras 5 filas
nba.head()
```

2. Selección de Columnas usando Corchetes `[]`

La forma recomendada y más segura para acceder a una columna de un DataFrame es pasando el nombre de la columna entre corchetes y comillas `df["nombre_columna"]`. Esto devuelve una estructura de tipo `Series`.

* **`nba["Name"]`**: Extrae la columna con los nombres de los jugadores.
* **`nba["Salary"]`**: Extrae la columna con los salarios de los jugadores.

```python
# Acceso a la columna 'Name'
nba["Name"]

# Acceso a la columna 'Salary'
nba["Salary"]
```

3. Selección de Columnas usando Sintaxis de Atributo (`.`)

Pandas también permite acceder a una columna como si fuera un atributo del DataFrame usando la notación de punto: `nba.Name`.

> **Nota:** Aunque es una forma rápida de escribir, solo funciona si el nombre de la columna no tiene espacios ni coincide con un método existente de Pandas (como `.shape`, `.count`, etc.).

```python
# Acceso a la columna 'Name' mediante notación de punto
nba.Name
```
