Optimización de Tipos de Datos (`category`) y Manipulación de Cadenas (`str`)

En este cuaderno se analiza el dataset de empleados aprendiendo a optimizar el uso de memoria convirtiendo columnas de texto repetitivo al tipo de dato `category`, así como la manipulación de cadenas de texto en Pandas utilizando el accesor `.str`.

1. Conexión a Drive, Carga e Inspección de Cardinalidad
Montamos Google Drive, cargamos el archivo `Employee.csv` y revisamos sus características iniciales. Usamos `.nunique()` para contar el número de valores únicos por columna e identificar variables categóricas candidatas a optimización.

```python
import pandas as pd
from google.colab import drive

# Montaje de Google Drive
drive.mount('/content/drive')

# Carga del conjunto de datos de empleados
emp = pd.read_csv("/content/drive/MyDrive/udmy/dataAnalisys/Employee.csv")

# Muestra inicial e información del sistema
emp.head()
emp.info()

# Conteo de valores únicos por columna
emp.nunique()
```

2. Conversión a Tipo de Dato Categórico (`astype("category")`)

Cuando una columna de texto tiene pocos valores únicos repetidos en muchas filas (baja cardinalidad, como departamento o posición), convertirla a `category` optimiza significativamente el uso de memoria en RAM.


* **`emp[["department", "position"]].astype("category")`**: Modifica el tipo de dato de tipo objeto/texto a tipo categórico.
* **`emp.info()`**: Permite verificar la reducción en el consumo de memoria del `DataFrame`.

```python
# Conversión de columnas de texto a tipo categórico
emp[["department", "position"]] = emp[["department", "position"]].astype("category")

# Verificación de la reducción de uso de memoria
emp.info()
```

3. Métodos de Cadenas nativos vs. Accesor `.str` en Pandas

En Python nativo, los métodos de cadenas se aplican a elementos individuales (`"happy".upper()`, `"PYTHON".lower()`).

Para aplicar métodos de cadenas de manera vectorizada sobre todos los elementos de una `Series` en Pandas, es necesario anteponer el accesor `.str`:
* **`emp["name"].str.lower()`**: Convierte todos los nombres a minúsculas.
* **`.str.title()`**: Encadena una segunda transformación para formatear cada nombre en mayúscula inicial (Formato Título).

```python
# Métodos de cadena nativos en Python (referencia)
"happy".upper()
"PYTHON".lower()

# Selección de la columna de nombres
emp["name"]

# Aplicación encadenada de transformaciones de texto en Pandas
emp["name"].str.lower().str.title()
```
