Restablecimiento del Orden por Índice (sort_index)
En este cuaderno se explora cómo devolver un DataFrame a su secuencia u orden original de filas mediante el método .sort_index(), tras haber realizado un ordenamiento previo por valores de columna.

1. Conexión a Drive, Carga de Datos y Ordenamiento por Valores
Se realiza el montaje de Google Drive, la lectura del conjunto de datos y la modificación permanente del orden de las filas según las columnas Team y Name posicionando los nulos al inicio.

```python
import pandas as pd
from google.colab import drive

# Montaje de Google Drive
drive.mount('/content/drive')

# Carga del conjunto de datos desde Excel
nba = pd.read_excel('/content/drive/MyDrive/udmy/dataAnalisys/Nba Teams.xlsx')

# Ordenamiento permanente por Equipo y Nombre con nulos al inicio
nba.sort_values(["Team", "Name"], na_position="first", inplace=True)

# Visualización del DataFrame reordenado
nba
```
