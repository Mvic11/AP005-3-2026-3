Modificación de Columnas en Pandas: Creación, Eliminación e Inserción

En este cuaderno se abordan las operaciones básicas para modificar las columnas de un `DataFrame`: añadir nuevas columnas con valores constantes, eliminar columnas existentes con `drop()` e insertar columnas en posiciones específicas con `insert()`.

1. Carga de Datos y Muestra Inicial
Se realiza la importación de `pandas`, la lectura del archivo `Nba Teams.xlsx` y la inspección básica de sus primeros registros.

```python
import pandas as pd

# Carga del conjunto de datos desde un archivo Excel
nba = pd.read_excel("Nba Teams.xlsx")

# Muestra de las primeras 5 filas
nba.head()
```

2. Creación de Nuevas Columnas con Asignación Directa

Para agregar una nueva columna al final del DataFrame, basta con asignarle un valor a una nueva clave en la sintaxis de corchetes: `df["nueva_columna"] = valor`.

* **`nba["Teams"] = " "`**: Crea la columna `Teams` llena de espacios en blanco.
* **`nba["nba Teams"] = " "`**: Crea la columna `nba Teams`.

> **Nota sobre notación de punto (`.`):**
> * Funciona correctamente con columnas de palabras simples como `nba.Teams`.
> * **No funciona** con columnas que contienen espacios como `nba.nba Teams` (daría error de sintaxis).

```python
# Creación de dos columnas vacías al final del DataFrame
nba["Teams"] = " "
nba["nba Teams"] = " "

# Verificación de cambios
nba.head()

# Prueba de acceso por atributo (notación de punto)
nba.Teams
```

3. Asignación de Valores Estáticos a una Columna

Podemos asignar un valor fijo a todas las filas de una columna seleccionada. En este caso, añadimos la columna `Sports` asignándole el valor constante `"basket"`.

```python
# Asignación de una nueva columna con valor constante
nba["Sports"] = "basket"

# Visualización de los cambios
nba.head()
```

4. Eliminación de Columnas (`drop`)

Para borrar una columna del DataFrame se utiliza el método `.drop()`:
* **`"Teams"`**: Nombre de la columna que se desea eliminar.
* **`axis = 1`**: Especifica que la operación se realiza sobre las columnas (eje vertical).
* **`inplace = True`**: Modifica el DataFrame directamente sin necesidad de reasignar la variable.

```python
# Eliminación permanente de la columna 'Teams'
nba.drop("Teams", axis=1, inplace=True)
```

5. Inserción de Columnas en una Posición Específica (`insert`)

A diferencia de la asignación directa que coloca las nuevas columnas al final, el método `.insert()` nos permite ubicar una columna en el índice de posición exacto que queramos:

* **Parámetro 1 (`2`)**: La posición de índice donde se insertará la columna (columna #3 en orden base cero).
* **`column = "Teams"`**: El nombre de la nueva columna.
* **`value = "Colombians"`**: El valor constante para llenar las filas.

```python
# Inserción de la columna 'Teams' en la tercera posición (índice 2)
nba.insert(2, column="Teams", value="Colombians")

# Visualización del DataFrame final
nba
```
