Optimización de Datasets: Tipos de Fechas (`datetime`) y Categóricos (`category`)

En este cuaderno se abordan técnicas clave para la optimización de memoria y la preparación de datos en Pandas, convirtiendo columnas de texto a tipos de datos nativos adecuados como fechas (`datetime`) y variables categóricas (`category`).

 1. Conexión a Drive, Carga e Inspección Inicial
Montamos Google Drive, importamos `pandas` y leemos el archivo `Instagram.csv` para inspeccionar la estructura inicial y el uso de memoria en RAM con `.info()`.

```python
import pandas as pd
from google.colab import drive

# Montaje de Google Drive
drive.mount('/content/drive')

# Carga del conjunto de datos de Instagram
insta = pd.read_csv('/content/drive/MyDrive/udmy/dataAnalisys/Instagram.csv')

# Inspección de los primeros registros y consumo de memoria
insta.head()
insta.info()
```

 2. Conversión a Formato Fecha (`pd.to_datetime`)

Por defecto, las columnas con fechas almacenadas en archivos CSV o Excel se cargan como cadenas de texto (`object`).

* **`pd.to_datetime()`**: Parsea las columnas de texto a objetos de fecha/hora de Pandas (`datetime64[ns]`), habilitando el uso de métodos temporales y reduciendo el costo computacional en filtros y cálculos de tiempo.

```python
# Conversión de las columnas 'reg date' y 'last login' a tipo datetime
insta["reg date"] = pd.to_datetime(insta["reg date"])
insta["last login"] = pd.to_datetime(insta["last login"])

# Verificación de la actualización en los tipos de datos
insta.info()
```

 3. Conversión a Categórico (`category`) e Inspección de Memoria

Para columnas de texto con pocos valores únicos (baja cardinalidad, como `gender`), la conversión a tipo `category` asigna un código entero interno a cada texto repetido, reduciendo considerablemente el uso de memoria RAM:

* **`insta["gender"].nunique()`**: Cuenta los valores únicos para confirmar si es una variable categórica adecuada.
* **`insta["gender"].astype("category")`**: Asigna el nuevo tipo de dato.
* **`insta["Country"]`**: Muestra la columna final del conjunto optimizado.

```python
# Verificación de la cantidad de valores únicos en la columna 'gender'
insta["gender"].nunique()

# Conversión permanente de 'gender' a tipo categórico
insta["gender"] = insta["gender"].astype("category")

# Verificación del impacto en el uso total de memoria RAM
insta.info()

# Visualización de la columna Country
insta["Country"]
```
