Conexión a Google Drive e Inspección de Valores Faltantes (Nulls)

En este cuaderno se conecta el entorno de ejecución de Google Colab con Google Drive para leer un archivo de Excel y se realiza un diagnóstico de la presencia de datos nulos o faltantes en el conjunto de datos.
1. Montaje de Google Drive e Importación de Datos
Para acceder a archivos almacenados en Google Drive dentro de Colab, se utiliza el módulo `drive` de `google.colab`. Posteriormente se lee el archivo de Excel deseado.

```python
import pandas as pd
from google.colab import drive

# Montaje de la unidad de Google Drive
drive.mount('/content/drive')

# Carga del archivo Excel a un DataFrame
nba = pd.read_excel('/content/drive/MyDrive/udmy/dataAnalisys/Nba Teams.xlsx')

# Visualización del DataFrame
nba
```

2. Diagnóstico del DataFrame (`nba.info()`)

El método `.info()` nos brinda un resumen técnico general donde podemos comparar el número total de filas contra el conteo de valores no nulos (*Non-Null Count*) de cada columna.

```python
# Muestra información general sobre filas, columnas y tipos de datos
nba.info()
```

3. Detección y Conteo de Valores Nulos (`isnull()` y `sum()`)

Para identificar exactamente en qué partes faltan datos:
* **`nba.isnull()`**: Devuelve una máscara booleana (`True` / `False`) del mismo tamaño del DataFrame, indicando `True` únicamente en las celdas donde el valor es nulo (`NaN`).
* **`nba.isnull().sum()`**: Suma todos los valores `True` de cada columna, devolviendo el **total exacto de registros faltantes/nulos por columna**.

```python
# Muestra la matriz booleana de valores nulos
nba.isnull()

# Conteo total de valores nulos por columna
nba.isnull().sum()
```
