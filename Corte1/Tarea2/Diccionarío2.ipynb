# MÉTODOS Y ACCESO EN DICCIONARIOS

Acceso a valores mediante claves y verificación de existencia

```python
# Sección para aprender a obtener un valor usando una clave.
# #### Get A Key
# # you can access the values in it by providing the key:

# Crea un diccionario con edificios como claves y sus alturas como valores.
building_heights = {
    "Burj Khalifa": 828,
    "Shanghai Tower": 632,
    "Abraj Al Bait": 601,
    "Ping An": 599,
    "Lotte World Tower": 554.5,
    "One World Trade": 541.3
}

print(building_heights["Burj Khalifa"]) # Busca la clave "Burj Khalifa" y muestra 828.
print(building_heights["Ping An"])       # Busca la clave "Ping An" y muestra 599.

# Crea un diccionario donde cada elemento zodiacal contiene una lista de signos.
zodiac_elements = {
    "water": ["Cancer", "Scorpio", "Pisces"],
    "fire": ["Aries", "Leo", "Sagittarius"],
    "earth": ["Taurus", "Virgo", "Capricorn"],
    "air": ["Gemini", "Libra", "Aquarius"]
}

print(zodiac_elements["earth"]) # Muestra la lista de signos asociados a tierra.
print(zodiac_elements["fire"])  # Muestra la lista de signos asociados al fuego.

## Get an Invalid Key
# Intenta obtener el valor de "Landmark 81", pero genera un KeyError porque la clave no existe.
# print(building_heights["Landmark 81"])

## One way to avoid this error is to first check if the key exists in the dictionary:
key_to_check = "Landmark 81" # Guarda la clave que queremos comprobar.

if key_to_check in building_heights: # Comprueba si la clave existe dentro del diccionario.
    print(building_heights["Landmark 81"]) # Si la clave existe, obtiene y muestra su valor.

# Agrega una nueva clave llamada "energy" con el valor "Not a Zodiac element".
zodiac_elements["energy"] = "Not a Zodiac element"

if "energy" in zodiac_elements: # Comprueba si la clave "energy" existe dentro del diccionario.
    print(zodiac_elements["energy"]) # Obtiene y muestra el valor asociado a "energy".
```

Acceso seguro mediante .get()

```python
## Safely Get a Key
building_heights = {
    "Burj Khalifa": 828,
    "Shanghai Tower": 632,
    "Abraj Al Bait": 601,
    "Ping An": 599,
    "Lotte World Tower": 554.5,
    "One World Trade": 541.3
}

# get() busca "Shanghai Tower" y devuelve su valor (632).
building_heights.get("Shanghai Tower")

# get() busca "My House"; como no existe, devuelve None sin lanzar error.
building_heights.get("My House")

###
user_ids = {
    "teraCoder": 100019,
    "pythonGuy": 182921,
    "samTheJavaMaam": 123112,
    "lyleLoop": 102931,
    "keysmithKeith": 129384
}

user_ids.get("teraCoder") # Busca el ID correspondiente al usuario "teraCoder".

if user_ids.get("teraCoder") == None: # Comprueba si "teraCoder" no tiene un ID.
    tc_id = 1000                      # Asigna 1000 como ID predeterminado.
else:
    tc_id = user_ids.get("teraCoder") # Guarda en tc_id el ID perteneciente a "teraCoder".

print(tc_id) # Muestra en pantalla el ID almacenado en tc_id.

if user_ids.get("superStackSmash") == None: # Comprueba si el usuario no existe.
    stack_id = 100000                        # Asigna 100000 como ID predeterminado.

print(stack_id) # Muestra en pantalla el valor almacenado en stack_id.
```

Eliminación de claves con .pop()

```python
### Delete a Key
# .pop() works to delete items from a dictionary, when you know the key value.

raffle = {
    223842: "Teddy Bear",
    872921: "Concert Tickets",
    320291: "Gift Basket",
    412123: "Necklace",
    298787: "Pasta Maker"
}

print(raffle.pop(320291, "No Prize")) # Elimina la clave 320291 y devuelve "Gift Basket".
print(raffle)                         # Muestra el diccionario tras la eliminación.

print(raffle.pop(100000, "No Prize")) # Intenta eliminar 100000; al no existir, devuelve "No Prize".
print(raffle)                         # El diccionario permanece igual.

print(raffle.pop(872921, "No Prize")) # Elimina la clave 872921 y devuelve "Concert Tickets".
print(raffle)                         # Muestra el diccionario actualizado.

available_items = {
    "health potion": 10,
    "cake of the cure": 5,
    "green elixir": 20,
    "strength sandwich": 25,
    "stamina grains": 15,
    "power stew": 30
}
health_points = 20 # Variable con puntos de vida iniciales.

# Elimina elementos sumando sus valores recuperados a health_points.
health_points += available_items.pop("stamina grains", 0) # Suma 15
health_points += available_items.pop("power stew", 0)     # Suma 30
health_points += available_items.pop("mystic bread", 0)   # No existe, suma 0

print(available_items) # Muestra los objetos que quedaron en el diccionario.
print(health_points)    # Muestra la cantidad final de puntos de vida.
```

Obtener todas las claves con .keys()

```python
## Get All Keys
test_scores = {
    "Grace": [80, 72, 90],
    "Jeffrey": [88, 68, 81],
    "Sylvia": [80, 82, 84],
    "Pedro": [98, 96, 95],
    "Martin": [78, 80, 78],
    "Dina": [64, 60, 75]
}

print(list(test_scores)) # Convierte las claves del diccionario en una lista y las muestra.

for student in test_scores.keys(): # Recorre todas las claves del diccionario.
    print(student)                 # Muestra el nombre de cada estudiante.

user_ids = {
    "teraCoder": 100019,
    "pythonGuy": 182921,
    "samTheJavaMaam": 123112,
    "lyleLoop": 102931,
    "keysmithKeith": 129384
}

num_exercises = {
    "functions": 10,
    "syntax": 13,
    "control flow": 15,
    "loops": 22,
    "lists": 19,
    "classes": 18,
    "dictionaries": 18
}

users = user_ids.keys()       # Guarda las claves de user_ids.
lessons = num_exercises.keys() # Guarda las claves de num_exercises.

print(users)   # Muestra las claves almacenadas en users.
print(lessons) # Muestra las claves almacenadas en lessons.
```

Obtener todos los valores con .values()

```python
## Get All Values
test_scores = {
    "Grace": [80, 72, 90],
    "Jeffrey": [88, 68, 81],
    "Sylvia": [80, 82, 84],
    "Pedro": [98, 96, 95],
    "Martin": [78, 80, 78],
    "Dina": [64, 60, 75]
}

for score_list in test_scores.values(): # Recorre todos los valores del diccionario.
    print(score_list)                    # Muestra la lista de calificaciones.

num_exercises = {
    "functions": 10,
    "syntax": 13,
    "control flow": 15,
    "loops": 22,
    "lists": 19,
    "classes": 18,
    "dictionaries": 18
}

total_exercises = 0 # Inicializa el acumulador.

for exercises in num_exercises.values(): # Recorre cada valor del diccionario.
    total_exercises += exercises         # Acumula la cantidad total de ejercicios.

print(total_exercises) # Muestra el total de ejercicios.
```

Obtener pares clave-valor con .items()

```python
## Get All Items
biggest_brands = {
    "Apple": 184,
    "Google": 141.7,
    "Microsoft": 80,
    "Coca-Cola": 69.7,
    "Amazon": 64.8
}

for company, value in biggest_brands.items(): # Recorre simultáneamente clave y valor.
    print(company + " has a value of " + str(value) + " billion dollars. ")

pct_women_in_occupation = {
    "CEO": 28,
    "Engineering Manager": 9,
    "Pharmacist": 58,
    "Physician": 40,
    "Lawyer": 37,
    "Aerospace Engineer": 9
}

for occupation, percentage in pct_women_in_occupation.items(): # Recorre clave y valor.
    print("Women make up " + str(percentage) + " percent of " + occupation + "s.")
```
