Agregar múltiples claves con update()

```python
## Add multiple keys
# Crea un diccionario con tres habitaciones y sus respectivas temperaturas.
sensors = {"living room": 21, "kitchen": 23, "bedroom": 20}
print("Before", sensors) # Muestra el diccionario sensors antes de agregar nuevos elementos.

# Es un comentario que explica que podemos agregar tres habitaciones nuevas.
# If we wanted to add 3 new rooms, we could use:

# update() agrega varias claves y valores al diccionario al mismo tiempo.
sensors.update({"pantry": 22, "guest room": 25, "patio": 34})
print("After", sensors) # Muestra el diccionario después de agregar las nuevas habitaciones.

###
# Crea un diccionario donde cada usuario tiene asociado un número de identificación.
user_ids = {"teraCoder": 9018293, "proProgrammer": 119238}
print(user_ids) # Muestra en pantalla los usuarios y sus respectivos identificadores.

# Agrega dos nuevos usuarios al diccionario utilizando update().
user_ids.update({"theLooper": 138475, "stringQueen": 85739})
print(user_ids) # Muestra el diccionario después de agregar los nuevos usuarios.
```

Comprensión de diccionarios (Dict Comprehensions) y zip()

```python
### Dict Comprehensions
# Explica que se tienen dos listas que se quieren combinar (estudiantes y sus alturas).
# Let’s say we have two lists that we want to combine into a dictionary...

# Crea una lista llamada names que contiene los nombres de cuatro estudiantes.
names = ['Jenny', 'Alexus', 'Sam', 'Grace']

# Crea una lista llamada heights que contiene las alturas correspondientes a los estudiantes.
heights = [61, 70, 67, 64]

# Explica que Python permite crear un diccionario utilizando una comprensión.
# Python allows you to create a dictionary using a dict comprehension...

zipStudents = zip(names, heights)     # zip() combina las listas names y heights formando pares de elementos.
print("zipStudents: ", zipStudents) # Muestra en pantalla el objeto generado por zip().

# Crea un diccionario usando los nombres como claves y las alturas como valores.
students = {key: value for key, value in zip(names, heights)}
# students is now {'Jenny': 61, 'Alexus': 70, 'Sam': 67, 'Grace': 64}
print(students) # Muestra en pantalla el diccionario students.

# Explica que zip() combina dos listas en pares y que esos pares pueden utilizarse para crear un diccionario.
# zip() combines two lists into an iterator of tuples...

drinks = ["espresso", "chai", "decaf", "drip"] # Crea una lista llamada drinks que contiene diferentes tipos de bebidas.
caffeine = [64, 40, 0, 120]                   # Crea una lista llamada caffeine que contiene la cantidad de cafeína.

zipped_drinks = zip(drinks, caffeine) # Combina las listas drinks y caffeine en pares utilizando zip().
print(zipped_drinks)                  # Muestra en pantalla el objeto creado por zip().

# Crea un diccionario donde cada bebida es una clave y su cantidad de cafeína es el valor.
drinks_to_caffeine = {key: value for key, value in zipped_drinks}
print(drinks_to_caffeine) # Muestra en pantalla el diccionario drinks_to_caffeine.
```

Ejercicio integrado y diccionarios anidados

```python
# Crea una lista llamada songs que contiene los nombres de seis canciones.
songs = ["Like a Rolling Stone", "Satisfaction", "Imagine", "What's Going On", "Respect", "Good Vibrations"]

# Crea una lista llamada playcounts que contiene la cantidad de reproducciones de cada canción.
playcounts = [78, 29, 44, 21, 89, 5]

# Crea un diccionario llamado plays combinando cada canción con su cantidad de reproducciones mediante zip() y dict comprehension.
plays = {key: value for key, value in zip(songs, playcounts)}
print(plays) # Muestra en pantalla el diccionario plays con las canciones y sus reproducciones.

plays.update({"Purple Haze": 1}) # Agrega la canción Purple Haze al diccionario con 1 reproducción.
plays.update({"Respect": 94})     # Modifica el valor de Respect y cambia sus reproducciones de 89 a 94.
print("After: ", plays)          # Muestra en pantalla el diccionario actualizado.

# Crea un diccionario llamado library que contiene otro diccionario llamado plays y otro diccionario vacío.
library = {"The Best Songs": plays, "Sunday Feelings": {}}
print(library) # Muestra en pantalla todo el contenido del diccionario library.
```
